```python

# Create bins
bins = [0, 10, 50, 100, 500, 1000, 5000, 10000, 50000]
labels = ['0–10', '10–50', '50–100', '100–500', '500–1K', '1K–5K', '5K–10K', '10K–50K']

# Assign bins
df['Amount_Bin'] = pd.cut(df['Amount'], bins=bins, labels=labels)

# Group by Amount_Bin and Class
amount_dist = df.groupby(['Amount_Bin', 'Class']).size().reset_index(name='Count')

# Pivot table for plotting
pivot_table = amount_dist.pivot(index='Amount_Bin', columns='Class', values='Count').fillna(0)
pivot_table.columns = ['Not Fraud', 'Fraud']

# Plot with log-scaled y-axis
pivot_table.plot(
    kind='bar',
    figsize=(10, 5),
    color=['skyblue', 'salmon'],
    log=True
)

plt.title("Binned Amount Distribution by Class (Log Scale)")
plt.xlabel("Transaction Amount Range")
plt.ylabel("Log(Transaction Count)")
plt.xticks(rotation=45)
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
if 'Amount_Bin' in df.columns:
    df.drop('Amount_Bin', axis=1, inplace=True)
```

```python
# --- Feature Engineering: Temporal & Velocity Features ---
# Sort by Time to ensure temporal sequence for rolling aggregations
df = df.sort_values(by='Time').reset_index(drop=True)

# Approximate cardholders/user profiles using PCA features V1 and V2
# Since the dataset is PCA-transformed and anonymized, we can construct proxy user profiles by rounding V1 and V2
# This groups transactions that share similar profile characteristics
df['Card_Proxy'] = df['V1'].round(1).astype(str) + "_" + df['V2'].round(1).astype(str)

# 1. Time delta between consecutive transactions for each proxy user (in seconds)
df['Time_Delta'] = df.groupby('Card_Proxy')['Time'].diff().fillna(-1)

# 2. Velocity features: transaction counts in a rolling window of past transactions
# A simple window is the last 5 transactions for this proxy card, computing the time span
df['Last_5_Tx_Time_Span'] = df.groupby('Card_Proxy')['Time'].diff(periods=5).fillna(-1)

# 3. Rolling average amount of the last 5 transactions per card proxy
df['Rolling_Mean_Amount_5'] = df.groupby('Card_Proxy')['Amount'].transform(lambda x: x.rolling(5, min_periods=1).mean())

# Drop the temporary Card_Proxy column to prevent over-reliance on clustered PCA components
df.drop('Card_Proxy', axis=1, inplace=True)

print("Feature engineering complete. Added 'Time_Delta', 'Last_5_Tx_Time_Span', and 'Rolling_Mean_Amount_5'.")
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset into train and test sets BEFORE scaling to prevent data leakage
X = df.drop('Class', axis=1)
y = df['Class']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("Train shape (raw):", X_train.shape)
print("Test shape (raw):", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Scale Amount, Time, and the new engineered temporal/velocity features
num_features = ['Amount', 'Time', 'Time_Delta', 'Last_5_Tx_Time_Span', 'Rolling_Mean_Amount_5']

X_train = X_train.copy()
X_test = X_test.copy()

for col in num_features:
    scaler = StandardScaler()
    X_train[f'{col}_scaled'] = scaler.fit_transform(X_train[[col]])
    X_test[f'{col}_scaled'] = scaler.transform(X_test[[col]])
    X_train.drop(col, axis=1, inplace=True)
    X_test.drop(col, axis=1, inplace=True)

print("Train shape (scaled):", X_train.shape)
print("Test shape (scaled):", X_test.shape)
```

```python
import lightgbm as lgb
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, average_precision_score

print("--- Training LightGBM with Custom Focal Loss ---")

# Define focal loss objective for binary classification
def focal_loss_objective(preds, train_data, alpha=0.25, gamma=2.0):
    labels = train_data.get_label()
    # Sigmoid activation to convert margins to probabilities
    p = 1.0 / (1.0 + np.exp(-preds))
    
    # Compute gradient (g) with respect to predictions
    g = np.where(
        labels == 1,
        -alpha * ((1 - p) ** gamma) * (gamma * p * np.log(p + 1e-15) + (1 - p)),
        (1 - alpha) * (p ** gamma) * (gamma * (1 - p) * np.log(1 - p + 1e-15) + p)
    )
    
    # Compute Hessian (h)
    h = p * (1 - p)
    return g, h

# Create LightGBM datasets
train_dataset = lgb.Dataset(X_train, label=y_train)
test_dataset = lgb.Dataset(X_test, label=y_test, reference=train_dataset)

# Set GBDT parameters (passing custom objective in params for LightGBM 4.0+ compatibility)
params = {
    'learning_rate': 0.05,
    'num_leaves': 31,
    'max_depth': 6,
    'random_state': 42,
    'verbose': -1,
    'objective': focal_loss_objective,
    'metric': 'auc'
}

# Train LightGBM with custom objective
focal_model = lgb.train(
    params,
    train_dataset,
    num_boost_round=500,
    valid_sets=[test_dataset],
    callbacks=[lgb.early_stopping(50, verbose=False)]
)

# Prediction and Evaluation
raw_preds_test = focal_model.predict(X_test, raw_score=True)
y_proba_focal = 1.0 / (1.0 + np.exp(-raw_preds_test))
y_pred_focal = (y_proba_focal > 0.5).astype(int)

print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_focal))
print("Classification Report:\n", classification_report(y_test, y_pred_focal, digits=4))
print(f"ROC-AUC: {roc_auc_score(y_test, y_proba_focal):.4f}")
print(f"Average Precision (PR-AUC): {average_precision_score(y_test, y_proba_focal):.4f}")
```
